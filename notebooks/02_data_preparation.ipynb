```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
folder = Path("../data/raw")

ranks = [
    "iron",
    "bronze",
    "silver",
    "gold",
    "platinum",
    "emerald",
    "diamond",
    "master",
    "grandmaster",
    "challenger"
]

dfs = []

for rank in ranks:
    temp = pd.read_csv(folder / f"{rank}.csv")
    temp["tier"] = rank.upper()
    dfs.append(temp)

df = pd.concat(dfs, ignore_index=True)

print("Dataset shape:", df.shape)
```

```python
df = df.dropna(subset=["role"]).copy()

print("Dataset shape after cleaning:", df.shape)
print("Missing values:", df.isnull().sum().sum())
```

```python
columns_to_remove = [
    "match_id",
    "puuid",
    "team_id",
    "game_creation"
]

df_model = df.drop(columns=columns_to_remove)

print(df_model.columns.tolist())
```

```python
features = [
    "win",
    "kills",
    "deaths",
    "assists",
    "cs",
    "gold",
    "damage",
    "vision",
    "game_duration",
    "dmg_share",
    "gold_share",
    "gold_per_min",
    "damage_per_min",
    "vision_per_min",
    "kill_participation"
]
```

```python
X = df[features]
y = df["tier"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Features used for Model 1:")
print(X.columns.tolist())

print("\nTarget distribution:")
print(y.value_counts())

print("\nMissing values in X:")
print(X.isnull().sum())
```
